```python
# PART A

import numpy as np
import pandas as pd

# Keras / TensorFlow
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Sklearn utilities
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

##############################################################################
# 1) READ THE DATA
##############################################################################
df = pd.read_csv("concrete_data.csv")  # Make sure the CSV is in the same folder

# Separate predictors (X) and target (y)
X = df.drop('Strength', axis=1)  # drop the Strength column
y = df['Strength']


##############################################################################
# 2) BUILD A FUNCTION TO CREATE OUR BASELINE MODEL
##############################################################################
def create_baseline_model():
    model = Sequential()
    # One hidden layer of 10 nodes, ReLU activation
    model.add(Dense(10, activation='relu', input_shape=(X.shape[1],)))
    # Output layer: single neuron for regression
    model.add(Dense(1))
    # Compile
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model


##############################################################################
# 3) REPEAT TRAINING/TESTING 50 TIMES & COLLECT MSEs
##############################################################################
num_runs = 50
mse_scores = []

for run in range(num_runs):
    # (i) Random train-test split (30% test)
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.30, random_state=None
    )

    # (ii) Build & train the model
    model = create_baseline_model()
    model.fit(X_train, y_train, epochs=50, verbose=0)

    # (iii) Evaluate the model on the test set
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_scores.append(mse)

##############################################################################
# 4) REPORT THE MEAN & STANDARD DEVIATION OF THE MSEs
##############################################################################
mean_mse = np.mean(mse_scores)
std_mse = np.std(mse_scores)

print("Part A (Baseline Model)")
print(f"Mean MSE over 50 runs: {mean_mse:.4f}")
print(f"Std of MSE over 50 runs: {std_mse:.4f}")
```
