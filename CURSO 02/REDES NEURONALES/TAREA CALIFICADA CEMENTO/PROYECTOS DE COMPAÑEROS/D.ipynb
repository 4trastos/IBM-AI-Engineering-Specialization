```
# PART D

import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

##############################################################################
# 1) READ & NORMALIZE
##############################################################################
df = pd.read_csv("concrete_data.csv")

X = df.drop('Strength', axis=1)
y = df['Strength']

X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_norm = (X - X_mean) / X_std


##############################################################################
# 2) BUILD A MODEL WITH 3 HIDDEN LAYERS
##############################################################################
def create_model_deep():
    model = Sequential()
    model.add(Dense(10, activation='relu', input_shape=(X_norm.shape[1],)))
    model.add(Dense(10, activation='relu'))
    model.add(Dense(10, activation='relu'))
    # Output layer
    model.add(Dense(1))
    # Compile
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model


##############################################################################
# 3) RUN 50 TIMES WITH 50 EPOCHS
##############################################################################
num_runs = 50
mse_scores = []

for run in range(num_runs):
    # Random split
    X_train, X_test, y_train, y_test = train_test_split(
        X_norm, y, test_size=0.30, random_state=None
    )

    # Train
    model = create_model_deep()
    model.fit(X_train, y_train, epochs=50, verbose=0)

    # Predict & measure MSE
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_scores.append(mse)

mean_mse = np.mean(mse_scores)
std_mse = np.std(mse_scores)

print("Part D (3 Hidden Layers, Normalized Data, 50 epochs)")
print(f"Mean MSE (50 runs): {mean_mse:.4f}")
print(f"Std of MSE (50 runs): {std_mse:.4f}")

# Compare to Part B’s mean MSE
```
