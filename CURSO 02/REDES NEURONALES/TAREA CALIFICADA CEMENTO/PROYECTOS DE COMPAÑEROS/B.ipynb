```python
# PART B

import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

##############################################################################
# 1) READ THE DATA
##############################################################################
df = pd.read_csv("concrete_data.csv")

X = df.drop('Strength', axis=1)
y = df['Strength']

##############################################################################
# 2) NORMALIZE THE DATA
##############################################################################
# Compute mean and std of each column in X
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)

# Apply normalization: (X - mean) / std
X_norm = (X - X_mean) / X_std


##############################################################################
# 3) FUNCTION TO CREATE MODEL (same as Part A)
##############################################################################
def create_model_norm():
    model = Sequential()
    model.add(Dense(10, activation='relu', input_shape=(X_norm.shape[1],)))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model


##############################################################################
# 4) TRAIN/TEST 50 TIMES & COLLECT MSEs
##############################################################################
num_runs = 50
mse_scores = []

for run in range(num_runs):
    # Random split
    X_train, X_test, y_train, y_test = train_test_split(
        X_norm, y, test_size=0.30, random_state=None
    )

    model = create_model_norm()
    model.fit(X_train, y_train, epochs=50, verbose=0)

    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_scores.append(mse)

mean_mse = np.mean(mse_scores)
std_mse = np.std(mse_scores)

print("Part B (Normalized Data)")
print(f"Mean MSE (50 runs): {mean_mse:.4f}")
print(f"Std of MSE (50 runs): {std_mse:.4f}")

# Compare this mean to Part A’s mean MSE
```
