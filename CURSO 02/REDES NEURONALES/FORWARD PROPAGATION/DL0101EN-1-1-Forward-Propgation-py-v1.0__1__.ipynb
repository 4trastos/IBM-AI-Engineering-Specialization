Para codificar una forma automática de hacer predicciones, generalicemos nuestra red. Aunque la red muestra una capa oculta, codificaremos la red para que tenga muchas capas ocultas. ## Initialize a Network


Comencemos definiendo formalmente la estructura de la red.

```python
n = 2 # number of inputs
num_hidden_layers = 2 # number of hidden layers
m = [2, 2] # number of nodes in each hidden layer
num_nodes_output = 1 # number of nodes in the output layer
```

Ahora que hemos definido la estructura de la red, sigamos adelante e inicialicemos los pesos y los sesgos en la red con números aleatorios. Para poder inicializar los pesos y los sesgos con números aleatorios, necesitaremos importar la biblioteca **Numpy**.

```python
import numpy as np # import the Numpy library

num_nodes_previous = n # number of nodes in the previous layer

network = {} # initialize network an an empty dictionary

# recorra cada capa e inicialice aleatoriamente los pesos y sesgos asociados con cada nodo
# observe cómo agregamos 1 a la cantidad de capas ocultas para incluir la capa de salida
for layer in range(num_hidden_layers + 1): 
    
    # determine name of layer
    if layer == num_hidden_layers:
        layer_name = 'output'
        num_nodes = num_nodes_output
    else:
        layer_name = 'layer_{}'.format(layer + 1)
        num_nodes = m[layer]
    
    # initialize weights and biases associated with each node in the current layer
    network[layer_name] = {}
    for node in range(num_nodes):
        node_name = 'node_{}'.format(node+1)
        network[layer_name][node_name] = {
            'weights': np.around(np.random.uniform(size=num_nodes_previous), decimals=2),
            'bias': np.around(np.random.uniform(size=1), decimals=2),
        }
    
    num_nodes_previous = num_nodes
    
print(network) # print network
```

¡Genial! Ahora, con el código anterior, podemos inicializar los pesos y los sesgos correspondientes a cualquier red de cualquier cantidad de capas ocultas y de nodos en cada capa. Pero pongamos este código en una función para que podamos ejecutarlo de manera repetitiva cada vez que queramos construir una red neuronal.

```python
def initialize_network(num_inputs, num_hidden_layers, num_nodes_hidden, num_nodes_output):
    
    num_nodes_previous = num_inputs # number of nodes in the previous layer

    network = {}
    
    # loop through each layer and randomly initialize the weights and biases associated with each layer
    for layer in range(num_hidden_layers + 1):
        
        if layer == num_hidden_layers:
            layer_name = 'output' # name last layer in the network output
            num_nodes = num_nodes_output
        else:
            layer_name = 'layer_{}'.format(layer + 1) # otherwise give the layer a number
            num_nodes = num_nodes_hidden[layer] 
        
        # initialize weights and bias for each node
        network[layer_name] = {}
        for node in range(num_nodes):
            node_name = 'node_{}'.format(node+1)
            network[layer_name][node_name] = {
                'weights': np.around(np.random.uniform(size=num_nodes_previous), decimals=2),
                'bias': np.around(np.random.uniform(size=1), decimals=2),
            }
    
        num_nodes_previous = num_nodes

    return network # return the network
```

## Compute Weighted Sum at Each Node


La suma ponderada en cada nodo se calcula como el producto escalar de las entradas y las ponderaciones más el sesgo. Por lo tanto, creemos una función llamada *compute_weighted_sum* que hace exactamente eso.


```python
def compute_weighted_sum(inputs, weights, bias):
    return np.sum(inputs * weights) + bias
```

## Compute Node Activation


Recordemos que la salida de cada nodo es simplemente una transformación no lineal de la suma ponderada. Para esta asignación, utilizamos funciones de activación. En este caso, utilizaremos la función sigmoidea como función de activación. Por lo tanto, definamos una función que tome una suma ponderada como entrada y devuelva la transformación no lineal de la entrada utilizando la función sigmoidea.


```python
def node_activation(weighted_sum):
    return 1.0 / (1.0 + np.exp(-1 * weighted_sum))
```

## Forward Propagation


La última parte de la creación de una red neuronal que pueda realizar predicciones es juntar todo. Por lo tanto, vamos a crear una función que aplique las funciones *compute_weighted_sum* y *node_activation* a cada nodo de la red y propague los datos hasta la capa de salida y genere una predicción para cada nodo de la capa de salida.


Para lograrlo, seguiremos el siguiente procedimiento:

1. Empezaremos con la capa de entrada como entrada de la primera capa oculta.
2. Calcularemos la suma ponderada en los nodos de la capa actual.
3. Calcularemos la salida de los nodos de la capa actual.
4. Configuraremos la salida de la capa actual como entrada de la siguiente capa.
5. Pasaremos a la siguiente capa de la red.
5. Repetiremos los pasos 2 a 4 hasta que calculemos la salida de la capa de salida.


```python
def forward_propagate(network, inputs):
    
    layer_inputs = list(inputs) # start with the input layer as the input to the first hidden layer
    
    for layer in network:
        
        layer_data = network[layer]
        
        layer_outputs = [] 
        for layer_node in layer_data:
        
            node_data = layer_data[layer_node]
        
            # compute the weighted sum and the output of each node at the same time 
            node_output = node_activation(compute_weighted_sum(layer_inputs, node_data['weights'], node_data['bias']))
            layer_outputs.append(np.around(node_output[0], decimals=4))
            
        if layer != 'output':
            print('The outputs of the nodes in hidden layer number {} is {}'.format(layer.split('_')[1], layer_outputs))
    
        layer_inputs = layer_outputs # set the output of this layer to be the input to next layer

    network_predictions = layer_outputs
    return network_predictions
```
