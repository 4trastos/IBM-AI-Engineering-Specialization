En este laboratorio, cargará un conjunto de datos de clientes, ajustará los datos y utilizará K-Nearest Neighbors para predecir un punto de datos. Pero ¿qué es **K-Nearest Neighbors**?

**K-Nearest Neighbors** es un algoritmo de aprendizaje supervisado, en el que los datos se "entrenan" con puntos de datos correspondientes a su clasificación. Para predecir la clase de un punto de datos determinado, tiene en cuenta las clases de los "K" puntos de datos más cercanos y elige la clase a la que pertenecen la mayoría de los "K" puntos de datos más cercanos como clase predicha.

### Aquí hay una visualización del algoritmo K-Vecinos más cercanos.
<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/images/KNN_Diagram.png">


En este caso, tenemos puntos de datos de las clases A y B. Queremos predecir cuál es la estrella (punto de datos de prueba). Si consideramos un valor k de 3 (los 3 puntos de datos más cercanos), obtendremos una predicción de la clase B. Sin embargo, si consideramos un valor k de 6, obtendremos una predicción de la clase A.

En este sentido, es importante tener en cuenta el valor de k. Con suerte, a partir de este diagrama, podrá hacerse una idea de lo que es el algoritmo de K vecinos más cercanos. Este algoritmo tiene en cuenta los 'K' vecinos más cercanos (puntos de datos) cuando predice la clasificación del punto de prueba.

### Conjunto de características

Definamos los conjuntos de características, X:

````python
df.columns
````

Para utilizar la biblioteca scikit-learn, tenemos que convertir el marco de datos de Pandas en una matriz Numpy:

````python
X = df[['region', 'tenure','age', 'marital', 'address', 'income', 'ed', 'employ','retire', 'gender', 'reside']] .values  #.astype(float)
X[0:5]
````

¿Cuales son nuestras etiquetas?

````python
y = df['custcat'].values
y[0:5]
````

## Normalize Data


La estandarización de datos proporciona a los datos una media cero y una varianza unitaria; es una buena práctica, especialmente para algoritmos como KNN, que se basa en la distancia de los puntos de datos:


````python
X = preprocessing.StandardScaler().fit(X).transform(X.astype(float))
X[0:5]
````

### Train Test Split

La precisión fuera de la muestra es el porcentaje de predicciones correctas que el modelo hace sobre datos con los que NO se ha entrenado. Si se entrena y se prueba el mismo conjunto de datos, lo más probable es que la precisión fuera de la muestra sea baja, debido a la probabilidad de que el modelo se sobreajuste.

Es importante que nuestros modelos tengan una precisión fuera de la muestra alta, porque el propósito de cualquier modelo, por supuesto, es hacer predicciones correctas sobre datos desconocidos. Entonces, ¿cómo podemos mejorar la precisión fuera de la muestra? Una forma es utilizar un enfoque de evaluación llamado División de entrenamiento/prueba.
La división de entrenamiento/prueba implica dividir el conjunto de datos en conjuntos de entrenamiento y prueba respectivamente, que son mutuamente excluyentes. Después de lo cual, se entrena con el conjunto de entrenamiento y se prueba con el conjunto de prueba.

Esto proporcionará una evaluación más precisa de la precisión fuera de la muestra porque el conjunto de datos de prueba no es parte del conjunto de datos que se ha utilizado para entrenar el modelo. Es más realista para los problemas del mundo real.


````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.2, random_state=4)
print ('Train set:', X_train.shape,  y_train.shape)
print ('Test set:', X_test.shape,  y_test.shape)
````

<h3>K nearest neighbor (KNN)</h3>


#### Import library


Classifier implementing the k-nearest neighbors vote.


````python
from sklearn.neighbors import KNeighborsClassifier
````

### Training

Comencemos el algoritmo con k=4 por ahora:


````python
k = 4
#Train Model and Predict  
neigh = KNeighborsClassifier(n_neighbors = k).fit(X_train,y_train)
neigh
````

### Predicción

Podemos utilizar el modelo para hacer predicciones sobre el conjunto de prueba:


````python
yhat = neigh.predict(X_test)
yhat[0:5]
````

### Evaluación de precisión
### Accuracy evaluation

En la clasificación de múltiples etiquetas, la **puntuación de precisión de la clasificación** es una función que calcula la precisión del subconjunto. Esta función es igual a la función jaccard_score. Básicamente, calcula en qué medida coinciden las etiquetas reales y las etiquetas previstas en el conjunto de prueba.


````python
from sklearn import metrics
print("Train set Accuracy: ", metrics.accuracy_score(y_train, neigh.predict(X_train)))
print("Test set Accuracy: ", metrics.accuracy_score(y_test, yhat))
````

#### ¿Qué sucede con otros K?

K en KNN es el número de vecinos más cercanos que se examinarán. Se supone que el usuario debe especificarlo. Entonces, ¿cómo podemos elegir el valor correcto para K?
La solución general es reservar una parte de los datos para probar la precisión del modelo. Luego, elija k = 1, use la parte de entrenamiento para el modelado y calcule la precisión de la predicción utilizando todas las muestras en su conjunto de prueba. Repita este proceso, aumente el k y vea qué k es el mejor para su modelo.

Podemos calcular la precisión de KNN para diferentes valores de k.


````python
Ks = 10
mean_acc = np.zeros((Ks-1))
std_acc = np.zeros((Ks-1))

for n in range(1,Ks):
    
    #Train Model and Predict  
    neigh = KNeighborsClassifier(n_neighbors = n).fit(X_train,y_train)
    yhat=neigh.predict(X_test)
    mean_acc[n-1] = metrics.accuracy_score(y_test, yhat)

    
    std_acc[n-1]=np.std(yhat==y_test)/np.sqrt(yhat.shape[0])

mean_acc
````

#### Grafique la precisión del modelo para un número diferente de vecinos.


````python
plt.plot(range(1,Ks),mean_acc,'g')
plt.fill_between(range(1,Ks),mean_acc - 1 * std_acc,mean_acc + 1 * std_acc, alpha=0.10)
plt.fill_between(range(1,Ks),mean_acc - 3 * std_acc,mean_acc + 3 * std_acc, alpha=0.10,color="green")
plt.legend(('Accuracy ', '+/- 1xstd','+/- 3xstd'))
plt.ylabel('Accuracy ')
plt.xlabel('Number of Neighbors (K)')
plt.tight_layout()
plt.show()
````

````python
print( "The best accuracy was with", mean_acc.max(), "with k=", mean_acc.argmax()+1) 
````
