#### Steps: 
**Step 1: Import Necessary Libraries**

Before you start, make sure to import the required libraries: TensorFlow and Keras. Keras is included within TensorFlow as `tensorflow.keras`. 


````python
# Install TensorFlow
!pip install tensorflow==2.16.2

import tensorflow as tf 
from tensorflow.keras.models import Model 
from tensorflow.keras.layers import Input, Dense 
import warnings
warnings.filterwarnings('ignore', category=UserWarning, module='tensorflow')
````

**Explanation:**

`!pip install tensorflow==2.16.2` installs the specified version of TensorFlow.

 `tensorflow` is the main library for machine learning in Python. 

 `Model` is used to create a model with the Functional API. 

 `Input` and `Dense` are types of layers that you will use in your model. 

 
**Step 2: Define the Input Layer**

You will define the input shape for your model. For simplicity, let's assume you are working with a dataset where each input is a vector of length 20. 

 


````python
input_layer = Input(shape=(20,))
print(input_layer)
````

**Explicación:**

`Input(shape=(20,))` crea una capa de entrada que espera vectores de entrada de longitud 20.

`print(input_layer)` muestra la información de la capa, lo que le ayuda a comprender el tipo de información que puede obtener sobre las capas.

**Paso 3: Agregar capas ocultas**

A continuación, agregará un par de capas ocultas a su modelo. Las capas ocultas ayudan al modelo a aprender patrones complejos en los datos.


````python
hidden_layer1 = Dense(64, activation='relu')(input_layer) 
hidden_layer2 = Dense(64, activation='relu')(hidden_layer1) 
````

**Explicación:**

`Dense(64, activation='relu')` crea una capa densa (completamente conectada) con 64 unidades y una función de activación ReLU.

Cada capa oculta toma la salida de la capa anterior como entrada.

**Paso 4: definir la capa de salida**

Por último, definirá la capa de salida. Suponga que está trabajando en un problema de clasificación binaria, por lo que la capa de salida tendrá una unidad con una función de activación sigmoidea.

````python
output_layer = Dense(1, activation='sigmoid')(hidden_layer2) 
````

**Explicación:**

`Dense(1, activation='sigmoid')` crea una capa densa con 1 unidad y una función de activación sigmoidea, adecuada para la clasificación binaria.

**Paso 5: Crear el modelo**

Ahora, creará el modelo especificando las capas de entrada y salida.


````python
model = Model(inputs=input_layer, outputs=output_layer)
model.summary()
````

**Explicación:**

`Model(inputs=input_layer, outputs=output_layer)` crea un modelo Keras que conecta la capa de entrada con la capa de salida a través de las capas ocultas.

`model.summary()` proporciona un resumen del modelo, mostrando las capas, sus formas y la cantidad de parámetros. Esto le ayuda a interpretar la arquitectura del modelo.

**Paso 6: compilar el modelo**

Antes de entrenar el modelo, debe compilarlo. Deberá especificar la función de pérdida, el optimizador y las métricas de evaluación.


````python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
````

**Explicación:**

`optimizer='adam'` especifica el optimizador Adam, una opción popular para entrenar redes neuronales.

`loss='binary_crossentropy'` especifica la función de pérdida para problemas de clasificación binaria.

`metrics=['accuracy']` le indica a Keras que evalúe el modelo utilizando la precisión durante el entrenamiento.

**Paso 7: Entrenar el modelo**

Ahora puede entrenar el modelo utilizando datos de entrenamiento. Para este ejemplo, supongamos que `X_train` son los datos de entrada de entrenamiento y `y_train` es la etiqueta correspondiente.


````python
# Example data (in practice, use real dataset) 

import numpy as np 
X_train = np.random.rand(1000, 20) 
y_train = np.random.randint(2, size=(1000, 1)) 
model.fit(X_train, y_train, epochs=10, batch_size=32) 
````

**Explicación:**

`X_train` e `y_train` son marcadores de posición para los datos de entrenamiento reales.

`model.fit` entrena el modelo para una cantidad específica de épocas y un tamaño de lote.

**Paso 8: evaluar el modelo**

Después del entrenamiento, puede evaluar el modelo en los datos de prueba para ver qué tan bien funciona.


````python
# Example test data (in practice, use real dataset) 

X_test = np.random.rand(200, 20) 
y_test = np.random.randint(2, size=(200, 1)) 
loss, accuracy = model.evaluate(X_test, y_test) 
print(f'Test loss: {loss}') 
print(f'Test accuracy: {accuracy}') 
````

**Explanation:**

`model.evaluate` computes the loss and accuracy of the model on test data. 

`X_test` and `y_test` are placeholders for your actual test data. 


### Dropout y Normalización por lotes

Antes de continuar con el ejercicio práctico, analicemos brevemente dos técnicas importantes que se utilizan a menudo para mejorar el rendimiento de las redes neuronales: **Capas de Dropout** y **Normalización por lotes**.

#### Capas de Dropout

El Dropout es una técnica de regularización que ayuda a evitar el sobreajuste en las redes neuronales. Durante el entrenamiento, Dropout establece aleatoriamente una fracción de unidades de entrada en cero en cada ciclo de actualización. Esto evita que el modelo dependa demasiado de neuronas específicas, lo que alienta a la red a aprender características más sólidas que se generalizan mejor a datos no vistos.

**Puntos clave:**
- El Dropout solo se aplica durante el entrenamiento, no durante la inferencia.
- La tasa de dropout es un hiperparámetro que determina la fracción de neuronas que se deben descartar.

#### Normalización por lotes

La normalización por lotes es una técnica que se utiliza para mejorar la estabilidad y la velocidad del entrenamiento de las redes neuronales. Normaliza la salida de una capa anterior al centrar y reescalar los datos, lo que ayuda a estabilizar el proceso de aprendizaje. Al reducir el cambio de covariable interna (los cambios en la distribución de las entradas de la capa), la normalización por lotes permite que el modelo utilice tasas de aprendizaje más altas, lo que a menudo acelera la convergencia.

**Puntos clave:**

- La normalización por lotes funciona normalizando las entradas de cada capa para que tengan una media de cero y una varianza de uno.
- Se aplica durante el entrenamiento y la inferencia, aunque su comportamiento varía ligeramente entre las dos fases.
- Las capas de normalización por lotes también introducen dos parámetros que se pueden aprender y que permiten que el modelo escale y desplace la salida normalizada, lo que ayuda a restaurar el poder de representación del modelo.


**Ejemplo de cómo agregar una capa Dropout en Keras:**

````python
from tensorflow.keras.layers import Dropout, Dense, Input
from tensorflow.keras.models import Model

# Define the input layer
input_layer = Input(shape=(20,))

# Add a hidden layer
hidden_layer = Dense(64, activation='relu')(input_layer)

# Add a Dropout layer
dropout_layer = Dropout(rate=0.5)(hidden_layer)

# Add another hidden layer after Dropout
hidden_layer2 = Dense(64, activation='relu')(dropout_layer)

# Define the output layer
output_layer = Dense(1, activation='sigmoid')(hidden_layer2)

# Create the model
model = Model(inputs=input_layer, outputs=output_layer)

# Summary of the model
model.summary()
````

**Ejemplo de cómo agregar normalización por lotes en Keras:**

````python
from tensorflow.keras.layers import BatchNormalization, Dense, Input
from tensorflow.keras.models import Model

# Define the input layer
input_layer = Input(shape=(20,))

# Add a hidden layer
hidden_layer = Dense(64, activation='relu')(input_layer)

# Add a BatchNormalization layer
batch_norm_layer = BatchNormalization()(hidden_layer)

# Add another hidden layer after BatchNormalization
hidden_layer2 = Dense(64, activation='relu')(batch_norm_layer)

# Define the output layer
output_layer = Dense(1, activation='sigmoid')(hidden_layer2)

# Create the model
model = Model(inputs=input_layer, outputs=output_layer)

# Summary of the model
model.summary()
````
